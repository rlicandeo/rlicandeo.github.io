```python
import numpy as np
from PIL import Image
img = Image.open('mohamed_hassan-silhouette-3113255_1280.png')
salmon_mask = np.array(img)
salmon_mask = 255 - salmon_mask

# salmon_mask = np.where(salmon_mask > 128, 255, 0).astype(np.uint8)
```

```python
fig = plt.figure(figsize=(14, 18))

plt.imshow(salmon_mask, cmap=plt.cm.gray, interpolation='bilinear')
plt.axis('off')
plt.show()
```

```python
# instantiate a word cloud object
rho_wc = WordCloud(background_color='white',
                    max_words=2000,
                      mask=salmon_mask,
                        stopwords=stopwords,
                        contour_width=2,
    contour_color="white"
)

# generate the word cloud
rho_wc.generate(text)

# display the word cloud
fig = plt.figure(figsize=(14, 18))

plt.imshow(rho_wc, interpolation='bilinear')
plt.axis('off')
plt.show()
```

```python


# Load and invert mask
img = Image.open("mohamed_hassan-silhouette-3113255_1920.png").convert("L")
mask = np.array(img)
mask = 255 - mask
mask = np.where(mask > 128, 255, 0).astype(np.uint8)

# Load text
with open("cv.txt", "r", encoding="utf-8") as f:
    text = f.read()

wc = WordCloud(
    background_color="white",
    mask=mask,
    contour_width=2,
    contour_color="black"
).generate(text)

plt.figure(figsize=(12, 12))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()
```
